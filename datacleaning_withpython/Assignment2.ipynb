<span  style="font-family:'New time Roman'; color:rgb(194,231,255); text-align:center;">
2. Descriptive statistics for numerical variables
</span>

<span  style="font-family:'New time Roman';">

- **Account Length (in months)**
- **Avg Monthly GB Download**
- **Age**
- **Total Charges**
</span>

```python
numerical_variables = ["Account Length (in months)", "Avg Monthly GB Download", "Age", "Total Charges"]
for col in numerical_variables:
    if col in churnDataset.columns:
        print(f"--- {col} ---")
        print(churnDataset[col].describe())
        print(f"Range: {churnDataset[col].max()-churnDataset[col].min()}")
        print(f"IQR: {churnDataset[col].quantile(0.25)-churnDataset[col].quantile(0.75)}")
        print(f"Mode: {churnDataset[col].mode()}")
        print("\n")
```

<span  style="font-family:'New time Roman'; color:rgb(194,231,255); text-align:center;">
3. Compare churned and non-churned customers
</span>
<span  style="font-family:'New time Roman';">

Based on Churn Label group :
- **Account Length (in months)**
- **Avg Monthly GB Download**
- **Age**
- **Total Charges**
</span>


```python
numerical_variables = ["Account Length (in months)", "Avg Monthly GB Download", "Age", "Total Charges"]

for col in numerical_variables:
    if col in churnDataset.columns:
        print(f"--- {col} ---")
        stats = churnDataset.groupby('Churn Label')[col].agg(['mean', 'median', 'std'])
        print(stats)
        print("\n")
```

<span  style="font-family:'New time Roman'; color:rgb(194,231,255); text-align:center;">
4. Frequency and relative-frequency tables
</span>

<span  style="font-family:'New time Roman';">

frequency tables using value_counts() for :
- Churn Label
- Contract Type
- Payment Method
- State
</span>

```python
catergorical_var = ["Churn Label","State" ,"Contract Type", "Payment Method"]
for col in catergorical_var:
    if col in churnDataset.columns:
        count =  churnDataset[col].value_counts()
        percentage = churnDataset[col].value_counts(normalize=True)
        table = pd.DataFrame({
            "Count": count,
            "Percentage": round(percentage*100, 2)
        })
        print(f"--------- {col} ------------")
        print(table.to_string())
        print(f"------------------------")
        print(f"The : {percentage.idxmax()}, is the maximum count with percentage = {round(percentage.max(),2)} %")
        print(f"------------------------")
        print("\n")
```

<span  style="font-family:'New time Roman'; color:rgb(194,231,255); text-align:center;">
5. Contingency tables and row percentages
</span>

<span  style="font-family:'New time Roman';">

Contingency tables involving Churn Label and three categorical variables:
- Contract Type
- Payment Method
- State
</span>

```python
catergorical_var = ["State" ,"Contract Type", "Payment Method"]
for col in catergorical_var:
    if col in churnDataset.columns:
        print(f"--- {col}/{'Churn Label'} ---")
        print(pd.crosstab(churnDataset["Churn Label"], churnDataset[col], margins=True))
        print("\n")
```

```python
catergorical_var = ["State" ,"Contract Type", "Payment Method"]
for col in catergorical_var:
    if col in churnDataset.columns:
        print(f"--- {col}/{'Churn Label'} ---")
        print(pd.crosstab(churnDataset["Churn Label"], churnDataset[col], normalize="index").round(3))
        print("\n")
```
