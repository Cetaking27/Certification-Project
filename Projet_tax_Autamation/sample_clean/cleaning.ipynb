```python
# transform column names, removing space 
df.columns = (
    df.columns
    .str.replace(' ', '_')
    .str.lower()
    .str.strip()
)
```

```python
#if i want to rename any colum label based on my requirement
df = df.rename(
    columns={
        "pack":"packs",
        "item_no":"items_number"
    }
)
df.head(0)
```

```python
#counting how many duplicate are in a dataset
numb_dup  = df.duplicated().sum()
numb_dup
```

```python
#if theres was duplicate what to do next

# first this code help you see the duplicated values by column 
df[df.duplicated(keep=False)].sort_values(by=list(df.columns)).head()

# drop all duplicate value and get the new dataset
df = df.drop_duplicates()

#chech the shape again
df.shape
```

```python
print(df.head(0))
print(df.dtypes)
```

```python
# how to convert numerical and categorical data

#create a copy of your dataset
df = df.copy()

#numeric value in dataset let say there is a missing value 
# total amont column
for column in ['sales_bottles']:
    df[column] = df[column].fillna(df[column].median())
    

# for categorical values:
# let say there's a missing value in columns, we'll fill them by "unknown"
categorical_cols = ['state_bottle_cost', 'state_bottle_retail', 'store_address','store_city']
df[categorical_cols] = df[categorical_cols].fillna('Unknown')
```

```python
# just to ckeck if the missing value actually was replaced
print(df.isnull().sum()[df.isnull().sum()>0])
```

```python
df.head(5)
```

```python
#how to map somes name in data to specifique name to
# standardizing all the categorical data to meet our requirement

for column in ['store_address','store_city']:
    df[column] = df[column].str.strip().str.upper()

store_city_map = {
    "BOONE":"Bovane City"
}
store_address_map = {
    "1111 8TH ST":"1111 8TH STREET"
}
df['store_city'] = df['store_city'].map(store_city_map).fillna(df['store_city'])
df['store_address']=df['store_address'].map(store_address_map).fillna(df['store_address'])
```

```python
df.head(5)
```

```python
#how to perform chain operation 
#this help remove some sign from data to transform them for calculation

#let say if sale_dollars colum has a dollars sign or currency sign

df['sales_dollars'] = (
    df['sales_dollars']
    .astype(str)
    .str.replace(r'[$@%!,]','',regex=True)
    .str.strip()
)
df['sales_dollars'] = pd.to_numeric(df['sales_dollars'], errors='coerce')
```

```python
df.head(4)
```

```python
#how to handle mix boolean values
# let create a dictionnary calls boolean map

bool_map = {
    "Yes": True, "yes":True, "Y":True, "y":True,
    "TRUE":True, "True":True, "true":True, "1":True, 
    1:True, True:True,
    "NO":False, "no":False, "No":False, "N":False,
    "n":False, 'False':False, "false":False, "FALSE":False,
    0:False, "0":False, False:False
}
df['sales_bottles'] = df['sales_bottles'].map(bool_map).fillna(False)
```

```python
#how to fix date in the same format 
df['ordered_on'] = pd.to_datetime(df['ordered_on'], dayfirst=False, errors='coerce')
failed = df['ordered_on'].isnull().sum()
```

```python
#how to extract some date attrubute
df['ordered_year'] = df['ordered_on'].dt.year
df['ordered_month'] = df['ordered_on'].dt.month
df['ordered_day_of_week'] = df['ordered_on'].dt.day_name()

print(df[['ordered_on','ordered_year', 'ordered_month','ordered_day_of_week']].head())
```
